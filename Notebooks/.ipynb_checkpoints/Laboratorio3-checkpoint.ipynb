```python
import pandas as pd
```

```python
import random
```

```python
#Crear un DataFrame con un 1000000 de registros
cantidad = 1000000
```

```python
#DatosAleatorios
influencers = ["BeachLifeMX","CancunVibes","MundoPlaya","CaribeTravel","SunsetInfluencer","ViajaConAna","LifeInCancun","ExploreMexico"]
comentarios = ["Genial","Fantastico","10/10","No me gusto","Mala experiencia"]
fecha = ["2026-05-01", "2026-05-02", "2026-05-03", "2026-05-04", "2026-05-05", "2026-05-06", "2026-05-07", "2026-05-07"]
tipos_post = ["Foto","Reel","Historia","Video","Carrusel"]
```

```python
dfInstagram = pd.DataFrame({
    "influencer" : [random.choice(influencers) for _ in range(cantidad)],
    "likes" : [random.randint(100,50000) for _ in range(cantidad)],
    "comentarios" : [random.choice(comentarios) for _ in range(cantidad)],
    "cantidad_comentarios" : [random.randint(1,5000) for _ in range(cantidad)],
    "fecha" : [random.choice(fecha) for _ in range(cantidad)],
    "tipo_post" : [random.choice(tipos_post) for _ in range(cantidad)]
})
```
